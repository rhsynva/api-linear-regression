```python
import numpy as np
from sklearn.linear_model import LinearRegression
from flask import Flask, request, jsonify
import threading
```

```python
#Squares of the houses
X = np.array([[40], [50], [65], [80], [100], [120]])
#The prices of the houses(AZN)
y = np.array([35000, 45000, 58000, 72000, 90000, 110000])
```

```python
#Training
```

```python
model = LinearRegression() #Creating a Linear Regression object
model.fit(X, y) #
```

```python
#3.INSTALLING THE FLASK API PROGRAM
```

```python
app = Flask(__name__)
```

```python
#4.DEFINING THE API ROUTE
```

```python
# We create an address here called '/predict' and only accept POST requests (sending data)
@app.route('/predict', methods=['POST'])
#We read JSON formatted data from the client (Request). We take the number corresponding to the 'field' key from the JSON.
#Since the model expects a 2D array, not a single number, we convert the incoming number to [[number]] format. We get the estimated value from our trained model.Since the result is inside the array, we select the first element [0] and convert it to an integer (int).
def predict_price():
    data = request.get_json()
    home_area = data['area']
    input_data = np.array([[home_area]])
    prediction = model.predict(input_data)
    calculated_price= int(prediction[0])
    #Convert the response back to JSON format and send it back to the client (Response)
    return jsonify({
        'status': 'successful',
        'input_area':home_area,
        'price': calculated_price
    })
```

```python
#5.STARTING THE SERVER IN THE BACKGROUND (Colab specific)
```

```python
#Normally, we would write app.run(), but we use threading to prevent the Colab cell from freezing.
def run_server():
    #We activate the server on localhost (127.0.0.1) and port 5000
    app.run(host='127.0.0.1', port=5000, debug=False, use_reloader=False)
#Creating a new thread and attach the server function to it.
server_thread = threading.Thread(target=run_server)
server_thread.start()
```
